```python
sns.set_context('paper', font_scale=1.5)

business_df = pd.read_csv('../datasets/yelp2019_business.csv')
business_feat_df = pd.read_csv(f'../datasets/yelp2019_business_categories.csv', index_col='business_id')

states = ['PA', 'FL', 'IN', 'MO', 'TN']

state_to_city = {'PA': 'Philadelphia', 'FL': 'Tampa', 'IN': 'Indianapolis', 'MO': 'Saint Louis', 'TN': 'Nashville'}

for state in states: #'Indianapolis', 'Nashville', 'Tucson', 'Tampa'

    curr_business_df = business_df[business_df.state == state]
    #curr_business_df.drop(columns=['Unnamed: 0'], inplace=True)
    # rename longitude and latitude columns as Longitude and Latitude
    curr_business_df.rename(columns={'longitude': 'Longitude', 'latitude': 'Latitude'}, inplace=True)
    # replace NaN values with 0
    curr_business_df['postal_code'].fillna(0, inplace=True)
    curr_business_df['postal_code'] = curr_business_df['postal_code'].astype(int).astype(str)
    postal_codes = curr_business_df['postal_code'].unique()

    curr_business_feat_df = business_feat_df.loc[curr_business_df.business_id]
    

    categories = np.array(curr_business_feat_df.columns.tolist())
    too_generic_categories = ['Restaurants', 'Food', 'Specialty Food', 'Shopping', 'Nightlife']
    filtered_categories = np.array([c for c in categories if c not in too_generic_categories])

    filtered_business_feat_df = curr_business_feat_df[filtered_categories]

    curr_business_df.categories = [c.split(', ') for c in curr_business_df.categories.values]

    macro_categories = ['Food', 'No Food']

    curr_business_df['macrocategory'] = curr_business_df.categories.apply(lambda x: 'Food' if any([c in x for c in ['Restaurants', 'Food', 'Specialty Food', 'Bars']]) else 'No Food')

    macro_cat_array = np.zeros((len(curr_business_feat_df), len(macro_categories)))

    for i, cat in enumerate(macro_categories):
        macro_cat_array[:, i] = curr_business_df.macrocategory.apply(lambda x: 1 if x == cat else 0).values
        
    frequency_categories = curr_business_feat_df[filtered_categories].sum().sort_values(ascending=False)
    most_frequent_categories = frequency_categories.index[:30]

    # load geojson "Zipcodes_Poly.geojson"
    import geopandas as gpd
    zipcodes = gpd.read_file(f'../datasets/{state}_zip_codes_geo.min.json')
    if 'ZCTA5CE10' in zipcodes.columns:
        # substitute 'ZCTA5CE10' with 'CODE' in zipcodes
        zipcodes.rename(columns={'ZCTA5CE10': 'CODE'}, inplace=True)
    elif 'ZIPCODE' in zipcodes.columns:
        # substitute 'ZIPCODE' with 'CODE' in zipcodes
        zipcodes.rename(columns={'ZIPCODE': 'CODE'}, inplace=True)

    # keep only the zipcodes in unique_pc
    zipcodes = zipcodes[zipcodes.CODE.isin(postal_codes)]

    # if city == 'Tampa':
    #     zipcodes = zipcodes[1:]

    #colors = ['red' if i==0 else 'white' for i in range(len(zipcodes))]

    # try to plot
    #fig, ax = plt.subplots(figsize=(5, 5))
    #zipcodes.plot(ax=ax, color=colors, edgecolor='black')

    for model_name in ['LightGCN', 'FeatLightGCN', 'FeatAndPCLightGCN']:
        curr_bus__out_embeddings = np.load(f'../results/yelp2019/{model_name}__bus__out_embeddings.npy')
        curr_bus__out_embeddings = curr_bus__out_embeddings[curr_business_df.index]

        np.random.seed(30398)

        n_pca_components = 8
        pca = PCA(n_components= n_pca_components)
        reduced_boe = pca.fit_transform(curr_bus__out_embeddings)


        # cluster all the businesses

        n_clusters = 2
        kmeans = KMeans(n_clusters=n_clusters, random_state=0, ).fit(reduced_boe)

        curr_business_df['kmeans_cluster'] = kmeans.labels_
        curr_business_df['kmeans_cluster'] = curr_business_df['kmeans_cluster'].astype('category')


        cluster_names = ['Cluster A', 'Cluster B']
        zipcodes_cluster_freq = pd.DataFrame(index=postal_codes)
        category_cluster_freq = pd.DataFrame(columns=macro_categories)

        for cluster in range(n_clusters):
            cluster_categories = filtered_business_feat_df.iloc[np.where(curr_business_df['kmeans_cluster'] == cluster)[0]].values

            cluster_category_distribution = np.sum(cluster_categories, axis=0)/np.sum(cluster_categories)

            cluster_zip_distribution = np.zeros(len(postal_codes))
            for i, postal_code in enumerate(postal_codes):
                cluster_zip_distribution[i] = np.sum(curr_business_df[curr_business_df['kmeans_cluster'] == cluster]['postal_code'] == postal_code)

            # count proportion of macro categories in the cluster
            cluster_macro_cat_distribution = np.sum(macro_cat_array[np.where(curr_business_df['kmeans_cluster'] == cluster)[0]], axis=0)

            # cluster_name = f'Cluster {macro_categories[np.argmax(cluster_macro_cat_distribution)]}'
            # if cluster_name in cluster_names:
            #     cluster_name = cluster_name + '2'
            # cluster_names.append(cluster_name)
            cluster_name = cluster_names[cluster]

            category_cluster_freq.loc[cluster_name, macro_categories] = cluster_macro_cat_distribution


            for i, postal_code in enumerate(postal_codes):
                if postal_code in zipcodes['CODE'].values:
                    zipcodes_cluster_freq.loc[postal_code, cluster_name] = cluster_zip_distribution[i]



        zipcodes_copy_for_plot = zipcodes.copy()
        for postal_code in postal_codes:
            if postal_code in zipcodes['CODE'].values:
                zipcodes_copy_for_plot.loc[zipcodes['CODE'] == str(postal_code), f'Most_freq_cluster'] = zipcodes_cluster_freq.loc[postal_code][1]/zipcodes_cluster_freq.loc[postal_code].sum()


        fig, axs = plt.subplots(1, 2, figsize=(14, 4.5))

        # plot the proportion of the business in cluster 1 for each postal code
        zipcodes_copy_for_plot.plot(ax=axs[0], edgecolor='black', facecolor='none')
        zipcodes_copy_for_plot.plot(column=f'Most_freq_cluster', cmap='magma', legend=True, ax=axs[0], vmin=0, vmax=1)
        axs[0].set_title(f'{cluster_names[1]} proportion for each postal code')
        axs[0].axis('off')
        if state == 'PA':
            axs[0].set_ylim(39.75, 40.6)
            axs[0].set_xlim(-76, -74.7)
        if state == 'FL':
            axs[0].set_ylim(27.3, 28.6)
            axs[0].set_xlim(-82.9, -81.4)
        if state == 'IN':
            axs[0].set_ylim(38.8, 40.6)
            axs[0].set_xlim(-87, -85)
        if state == 'MO':
            axs[0].set_ylim(38.2, 39)
            axs[0].set_xlim(-90.8, -90)
        if state == 'TN':
            axs[0].set_ylim(35.6, 36.85)
            axs[0].set_xlim(-87.9, -85.6)


        # plot an histogram of category_cluster_freq where x is a cluster, with stacked bars for each category
        # orient the xlabel in horizontal
        category_cluster_freq = category_cluster_freq.div(category_cluster_freq.values.sum())
        category_cluster_freq.plot(kind='bar', stacked=True, ax=axs[1], rot=0) 
        axs[1].set_title('Macro category distribution in each cluster')
        fig.tight_layout()

        #fig.suptitle(f'{model_name} - {city}')

        plt.show()

        fig.savefig(f'../results/figures/{state_to_city[state]}_{cluster_names[1]}_{model_name}_distribution.pdf')


        np.random.seed(30398)

        n_sub_clusters = 4

        zipcodes_copy_for_plot = zipcodes.copy()

        

        fig, axs = plt.subplots(1, n_clusters, figsize=(14, 4.5))
        for cluster_to_cluster in range(n_clusters):
            zipcodes_cluster_freq = pd.DataFrame(index=postal_codes, columns=range(n_sub_clusters))

            bus_cluster_idx = np.where(curr_business_df['kmeans_cluster'] == cluster_to_cluster)[0]
            bus_cluster_index = curr_business_df.iloc[bus_cluster_idx].index

            bus_cluster = reduced_boe[bus_cluster_idx]

            clustering = KMeans(n_clusters=n_sub_clusters, random_state=0, ).fit(bus_cluster)
            clustering.fit(bus_cluster)

            curr_business_df.loc[bus_cluster_index, 'kmeans_sub_cluster'] = clustering.labels_

            for cluster in range(n_sub_clusters):

                # cluster_categories = filtered_business_feat_df.iloc[(np.where((business_df['kmeans_cluster'] == cluster_to_cluster) & (business_df['kmeans_sub_cluster'] == cluster))[0])].values
                # cluster_category_distribution = np.sum(cluster_categories, axis=0)/np.sum(cluster_categories)

                cluster_zip_distribution = np.zeros(len(postal_codes))
                for i, postal_code in enumerate(postal_codes):
                    cluster_zip_distribution[i] = np.sum(curr_business_df[(curr_business_df['kmeans_cluster'] == cluster_to_cluster) & (curr_business_df['kmeans_sub_cluster'] == cluster)]['postal_code'] == postal_code)#/np.sum(business_df['kmeans_sub_cluster'] == cluster)

                # most_frequent_categories_in_cluster = cluster_category_distribution.argsort()[::-1][:5]
                # most_frequent_postal_codes_in_cluster = cluster_zip_distribution.argsort()[::-1][:5]

                for i, postal_code in enumerate(postal_codes):
                    if str(postal_code) in zipcodes['CODE'].values:
                        zipcodes_cluster_freq.loc[postal_code, cluster] = cluster_zip_distribution[i]

            

            for postal_code in postal_codes:
                if str(postal_code) in zipcodes['CODE'].values:
                    clusters_freq = zipcodes_cluster_freq.loc[postal_code]
                    zipcodes_copy_for_plot.loc[zipcodes['CODE'] == str(postal_code), f'Most_freq_cluster'] = clusters_freq.idxmax()

            zipcodes_copy_for_plot = zipcodes_copy_for_plot.astype({'Most_freq_cluster': 'int64'})

            zipcodes_copy_for_plot.plot(ax=axs[cluster_to_cluster], edgecolor='black', facecolor='none')

            # colors = sns.color_palette('tab10', n_sub_clusters)
            # colors = np.array(colors)
            # colors = colors[zipcodes_copy_for_plot['Most_freq_cluster'].values]

            zipcodes_copy_for_plot.plot(column=f'Most_freq_cluster', cmap='tab20', categorical=True, legend=True, ax=axs[cluster_to_cluster])

            axs[cluster_to_cluster].set_title(f'Most present subcluster of {cluster_names[cluster_to_cluster]}')
            #fig.suptitle(f'{model_name} - {city}')
            axs[cluster_to_cluster].axis('off')
            leg1 = axs[cluster_to_cluster].get_legend()
            for ea in leg1.legend_handles:
                ea.set_marker('s')
                ea.set_markeredgecolor('k')
                ea.set_markeredgewidth(0.7)
            if state == 'PA':
                axs[cluster_to_cluster].set_ylim(39.75, 40.6)
                axs[cluster_to_cluster].set_xlim(-76, -74.7)
            if state == 'FL':
                axs[cluster_to_cluster].set_ylim(27.3, 28.6)
                axs[cluster_to_cluster].set_xlim(-82.9, -81.4)
            if state == 'IN':
                axs[cluster_to_cluster].set_ylim(38.8, 40.6)
                axs[cluster_to_cluster].set_xlim(-87, -85)
            if state == 'MO':
                axs[cluster_to_cluster].set_ylim(38.2, 39)
                axs[cluster_to_cluster].set_xlim(-90.8, -90)
            if state == 'TN':
                axs[cluster_to_cluster].set_ylim(35.6, 36.85)
                axs[cluster_to_cluster].set_xlim(-87.9, -85.6)
            
        fig.tight_layout()
        fig.savefig(f'../results/figures/{state_to_city[state]}_subcluster_distrib_{model_name}_distribution.pdf')


```

## Hierarchical clustering

```python
from scipy.cluster.hierarchy import dendrogram

def plot_dendrogram(model, **kwargs):
    # Create linkage matrix and then plot the dendrogram

    # create the counts of samples under each node
    counts = np.zeros(model.children_.shape[0])
    n_samples = len(model.labels_)
    for i, merge in enumerate(model.children_):
        current_count = 0
        for child_idx in merge:
            if child_idx < n_samples:
                current_count += 1  # leaf node
            else:
                current_count += counts[child_idx - n_samples]
        counts[i] = current_count

    linkage_matrix = np.column_stack(
        [model.children_, model.distances_, counts]
    ).astype(float)

    # Plot the corresponding dendrogram
    dendrogram(linkage_matrix, **kwargs, orientation='left')
```

```python
clustering = AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward', compute_distances=True)
clustering.fit(curr_bus__out_embeddings)


plot_dendrogram(clustering, truncate_mode="level", p=1)
```

```python

from sklearn.cluster import AgglomerativeClustering

# for each cluster in the hierarchical clustering, get the frequency of the categories
# and the distribution across the postal codes



n_sub_clusters = 4

zipcodes_copy_for_plot = zipcodes.copy()

zipcodes_cluster_freq = pd.DataFrame(index=postal_codes, columns=range(n_sub_clusters*2))

for cluster_to_cluster in range(n_clusters):

    bus_cluster_idx = np.where(curr_business_df['hierarchical_cluster'] == cluster_to_cluster)[0]

    bus_cluster = curr_bus__out_embeddings[bus_cluster_idx]

    clustering = AgglomerativeClustering(n_clusters=n_sub_clusters, metric='euclidean', linkage='ward', compute_distances=True)
    clustering.fit(bus_cluster)


    curr_business_df.loc[bus_cluster_idx, 'hierarchical_sub_cluster'] = clustering.labels_

    

    for cluster in range(n_sub_clusters):

        cluster_categories = filtered_business_feat_df.iloc[(np.where((curr_business_df['hierarchical_cluster'] == cluster_to_cluster) & (curr_business_df['hierarchical_sub_cluster'] == cluster))[0])].values
        cluster_category_distribution = np.sum(cluster_categories, axis=0)/np.sum(cluster_categories)

        cluster_zip_distribution = np.zeros(len(postal_codes))
        for i, postal_code in enumerate(postal_codes):
            cluster_zip_distribution[i] = np.sum(curr_business_df[(curr_business_df['hierarchical_cluster'] == cluster_to_cluster) & (curr_business_df['hierarchical_sub_cluster'] == cluster)]['postal_code'] == postal_code)#/np.sum(business_df['hierarchical_sub_cluster'] == cluster)

        most_frequent_categories_in_cluster = cluster_category_distribution.argsort()[::-1][:5]
        most_frequent_postal_codes_in_cluster = cluster_zip_distribution.argsort()[::-1][:5]

        print(f'Cluster {cluster} - {filtered_categories[most_frequent_categories_in_cluster]}')
        print(f'locations: {postal_codes[most_frequent_postal_codes_in_cluster]}')
        print(f'size: {np.sum((curr_business_df['hierarchical_cluster'] == cluster_to_cluster) & (curr_business_df['hierarchical_sub_cluster'] == cluster))}')

        for i, postal_code in enumerate(postal_codes):
            if str(postal_code) in zipcodes['CODE'].values:
                zipcodes_cluster_freq.loc[postal_code, n_sub_clusters*cluster_to_cluster + cluster] = cluster_zip_distribution[i]
    
        cluster_macro_cat_distribution = np.sum(macro_cat_array[np.where((curr_business_df['hierarchical_cluster'] == cluster_to_cluster) & (curr_business_df['hierarchical_sub_cluster'] == cluster))[0]], axis=0)
        print(f'Macro category: {macro_categories[np.argmax(cluster_macro_cat_distribution)]}\nwith {np.max(cluster_macro_cat_distribution/np.sum(macro_cat_array, axis=0))*100:.2f}% of the businesses of the macro category in the cluster')
        

    fig, axs = plt.subplots(1, 1, figsize=(7, 5), sharey=True)

    for postal_code in postal_codes:
        if str(postal_code) in zipcodes['CODE'].values:
            zipcodes_copy_for_plot.loc[zipcodes['CODE'] == str(postal_code), f'Most_freq_cluster'] = int(zipcodes_cluster_freq.loc[postal_code, cluster_to_cluster*n_sub_clusters:(cluster_to_cluster+1)*n_sub_clusters].idxmax())

    zipcodes_copy_for_plot.plot(column=f'Most_freq_cluster', cmap='viridis', categorical=True, legend=True, ax=axs)

    axs.set_title(f'Most present subcluster of cluster {cluster_names[cluster_to_cluster]}')
    axs.axis('off')

fig, axs = plt.subplots(1, 1, figsize=(7, 5), sharey=True)

for postal_code in postal_codes:
    if str(postal_code) in zipcodes['CODE'].values:
        most_freq_subcluster = zipcodes_cluster_freq.loc[postal_code].idxmax()
        cluster_label = most_freq_subcluster//n_sub_clusters
        sub_cluster_label = most_freq_subcluster%n_sub_clusters

        zipcodes_copy_for_plot.loc[zipcodes['CODE'] == str(postal_code), f'Most_freq_cluster'] = str(cluster_label) + '_' + str(sub_cluster_label)
        print(f'{postal_code}: predominant cluster {str(cluster_label) + '.' + str(sub_cluster_label)} with freq. {zipcodes_cluster_freq.loc[postal_code].max()/zipcodes_cluster_freq.loc[postal_code].sum()}')

zipcodes_copy_for_plot['Most_freq_cluster'] = zipcodes_copy_for_plot['Most_freq_cluster'].astype('category')
zipcodes_copy_for_plot.plot(column=f'Most_freq_cluster', cmap='viridis',  legend=True, ax=axs)

axs.set_title(f'Most present subcluster')
axs.axis('off')

```
